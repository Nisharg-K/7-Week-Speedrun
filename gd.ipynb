```python
import random

# Dataset
x = [1, 2, 3, 4, 5]
y = [3, 5, 7, 9, 11]

# Parameters
w = 0.5
b = 0.1

learning_rate = 0.01
epochs = 100

for epoch in range(epochs):

    loss = 0

    for i in range(len(x)):

        # Forward pass for ONE sample
        y_pred = w * x[i] + b

        error = y_pred - y[i]

        loss += error ** 2

        # Gradients from ONE sample only
        dw = 2 * error * x[i]
        db = 2 * error

        # Immediate update
        w = w - learning_rate * dw
        b = b - learning_rate * db

    loss /= len(x)

    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Loss = {loss:.4f}")

print("Final w:", w)
print("Final b:", b)
```
